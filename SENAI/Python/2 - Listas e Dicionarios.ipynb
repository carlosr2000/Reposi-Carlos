```python
# @title
produtos = ['Teclado','Mouse','Monitor','Headset','Webcam']
while True:

  print('''COMANDOS :
1 - Adicionar produto
2 - Remover produto
3 - Consultar produtos
4 - Sair''')
  opcao = int(input('Digite a opcao: '))
  if opcao == 1:
    produtos.append(input('Digite o nome do produto: '))
  elif opcao == 2:
    prod = input('Digite o nome do produto que deseja remover: ')
    if prod not in produtos:
      print('Produto não encontrado')
    else:
     produtos.remove(prod)
  elif opcao == 3:
    print(produtos)
  elif opcao == 4:
    print('Saindo do sistema.')
    break
  else:
    print('Opcão incorreta!')
```

```python
# @title
# Atividade 2 — Cadastro de Produto
# Crie um dicionário chamado produto contendo:
# Nome
# Categoria
# Preço
# Estoque
# Exemplo:
# Produto: Mouse
# Categoria: Informática
# Preço: R$ 80,00
# Estoque: 15
# O programa deve:
# Criar o dicionário.
# Exibir os dados do produto.
# Calcular o valor total do estoque.
# Fórmula:
# preço × quantidade em estoque
produtos = []
while True:
  print('''
  Seleciona o que deseja fazer:
  1 - Criar produto:
  2 - Exibir os dados do produto.
  3 - Calcular o valor total do estoque
  ''')
  opcao = int(input('Digite a opção: '))
  if opcao == 1:
    nome = input('Nome do Produto: ')
    categoria = input('Categoria do produto: ')
    estoque = int(input('Estoque: '))
    preço = int(input('Preço do produto: '))
    produto = {
       'nome': nome,
       'categoria': categoria,
       'estoque' : estoque,
       'preço' : preço}
    produtos.append(produto)
  if opcao == 2:
    print(produtos)
  if opcao == 3:
    for valor in produtos:
      valor_estoque = valor['preço'] * valor['estoque']
    print(valor_estoque)
```

```python
def analisar_numero(n:int) ->int:
  print(f'{n} analisado.')
  if n == 0:
    print('é zero')
  elif n > 0:
    print('é positivo')
  else:
    print('é negativo')

  if n % 2 == 0:
    print('e par\n')
  else:
    print('e impar\n')

analisar_numero(5)

analisar_numero(0)
analisar_numero(-5)

```
