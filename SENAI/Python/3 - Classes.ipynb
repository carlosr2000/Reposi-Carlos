```python
class Pessoa:
  def __init__(self, nome, idade):
    self.nome = nome
    self.idade = idade
  def boas_vindas(self):
    print(f'Bem vindo {self.nome}!')

pesso1 = Pessoa('Dado', 15)
pesso2 = Pessoa('Dardo',20)
pesso1.boas_vindas()
```

```python
class Produtos:
  def __init__(self, produto, preço , estoque):
    self.produto = produto
    self.preço = preço
    self.estoque = estoque
  def vender(self,quantidade):
    if quantidade > 0 and quantidade <= self.estoque:
      self.estoque -= quantidade
      print('Venda realizada com sucesso!')
    else:
      print('Quantidade acima do estoque!')
  def exibir_dados(self):
    print(f'''
    Produto : {self.produto}
    Preço de R$:{self.preço}
    Quantidade em estoque:{self.estoque}''')
  def valor_total(self):
    total = self.preço * self.estoque
    print(f'{self.produto} tem valor de estoque de R${total}!')

dado = Produtos('Dado',20,100)
dado.exibir_dados()
dado.valor_total()
```

```python
class Aluno():
  def __init__(self, nome, idade, curso, nota1, nota2):
    self.nome = nome
    self.idade = int(idade)
    self.curso = curso
    self.nota1 = float(nota1)
    self.nota2 = float(nota2)

  def calcular_media(self):
    media = ( self.nota1 + self.nota2 )/2
    print(f'A média é : {media}')

  def verificar_condicao(self):
    if (self.nota1 + self.nota2)/2 >= 7:
      print('Aluno aprovado!')
    else:
      print('Aluno Reprovado!')
aluno1 = Aluno('Carlos' , 20, 'DS', 10, 7)
aluno1.calcular_media()
aluno1.verificar_condicao()
```

```python
class Carro():
  def __init__(self, marca , modelo , ano, velocidade_atual):
    self.marca = marca
    self.modelo = modelo
    self.ano = ano
    self.velocidade_atual = velocidade_atual
  def acelerar(self, acelero):
    if acelero > 0:
      self.velocidade_atual += acelero
      print(f'Velocidade atual de : {self.velocidade_atual}')
    else:
      print('Acelerou pouco dms!')
  def frear(self, freio):
    while True:
      if self.velocidade_atual - freio <= 0:
        print('Freio irá parar o carro digite um freio menor!')
        freio = int(input())
      else:
        self.velocidade_atual-=freio
        print(f'A velocidade atual é de : {self.velocidade_atual}')
        break
  def dados(self):
    print(f'''
    {self.marca}: Marca
    {self.modelo} : Modelo
    {self.ano}: Ano de Fabricação
    {self.velocidade_atual} : velocidade atual do veiculo
     ''')
carro1 = Carro('Honda', 'sla', 2000, 100)


while True:
  print('''1 para freio
  2 para acelero
  3 pra exibir:
  4 pra parar
  ''')
  print(carro1.velocidade_atual)
  print('='*20)
  opcao = input('Digite a opcao: ')
  if opcao == '1':
    f = int(input('O quanto de freio: '))
    carro1.frear(f)

  if opcao == '2':
    a = int(input('O quanto de acelero: '))
    carro1.acelerar(a)

  if opcao == '3':
    carro1.dados()

  if opcao == '4':
    print('cabo')
    break
```
