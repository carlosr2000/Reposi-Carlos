```python
#DESAFIO 01 — Sistema de Cadastro e Análise de Alunos
n = int(input('Insira quantidade de alunos que sera inserido: '))
contador_de_aluno = 0
for _ in range(n):
  contador_de_aluno += 1
  nome = input('Nome do aluno: ')
  idade = input('Idade do aluno: ')
  cidade = input('Cidade do aluno: ')
  nota_1 = float(input('Nota da primeira avaliação: '))
  nota_2 = float(input('Nota da segunda avaliação: '))
  media = (nota_1 + nota_2) / 2
  if  media >= 7:
    print(f'Aluno {contador_de_aluno} Nome:{nome} - Idade:{idade}\nCidade:{cidade} - Media:{media}\nSituação: Aprovado\n =====')
  elif media >= 5:
    print(f'Aluno {contador_de_aluno} Nome:{nome} - Idade:{idade}\nCidade:{cidade} - Media:{media}\nSituação: Recuperação\n======')
  else:
    print(f'Aluno {contador_de_aluno} Nome:{nome} - Idade:{idade}\nCidade:{cidade} - Media:{media}\nSituação: Reprovado\n======')
#Média >= 7       → Aprovado
#Média >= 5       → Recuperação
#Média < 5        → Reprovado

```

```python
saldo = float(input('Insira seu saldo: '))

while True:
  print('''
  $$$$$$$$$$$$$$$$$$$$$$$$$$$
      CAIXA ELETRÔNICO
  $$$$$$$$$$$$$$$$$$$$$$$$$$$

  1 - Consultar saldo
  2 - Depositar
  3 - Sacar
  4 - Sair ''')
  opcao = int(input('Insira a opção: '))
  if opcao == 1:
    print(f'Seu saldo é de R${saldo}')
  elif opcao == 2:
    deposito = float(input('Insira o valor do seu deposito: '))
    while deposito <= 0:
      print('Valor abaixo do permitido. ')
      deposito = float(input('Insira um valor válido: '))
    saldo += deposito
    print(f'Seu saldo é de R${saldo}')
  elif opcao == 3:
    saque = float(input('Insira o valor do seu saque: '))
    if saque > saldo:
      print(f'Saldo insuficiente você possuí R${saldo}')
    elif saque <= 0:
      print(f'Saque inferior ao permitido!')
    else:
      saldo -= saque
      print(f'Saque de R${saque} realizado com sucesso e seu saldo é de R${saldo}')
  elif opcao == 4:
    print('''
$$$$$$$$$$$$$$$$$$$$$$$$$$$
Obrigado por usar o sistema
$$$$$$$$$$$$$$$$$$$$$$$$$$$''')
    break

  else:
    print('Opção invalida.')
```

```python
numeros = []
pares = []
impares = []

for v in range(10):
  valor = int(input(f"Digite o {v+1} º numero: "))
  numeros.append(valor)
for n in numeros:
  if n % 2 == 0:
    pares.append(n)
  else:
    impares.append(n)

print("Numeros Pares",pares)
print("Numeros Impares",impares)
```

```python
produtos = ['Teclado','Mouse','Monitor','Headset','Webcam']
while True:

  print('''COMANDOS :
1 - Adicionar produto
2 - Remover produto
3 - Consultar produtos
4 - Sair''')
  opcao = int(input('Digite a opcao: '))
  if opcao == 1:
    produtos.append(input('Digite o nome do produto: '))
  elif opcao == 2:
    prod = input('Digite o nome do produto que deseja remover: ')
    if prod not in produtos:
      print('Produto não encontrado')
    else:
     produtos.remove(prod)
  elif opcao == 3:
    print(produtos)
  elif opcao == 4:
    print('Saindo do sistema.')
    break
  else:
    print('Opcão incorreta!')
```
